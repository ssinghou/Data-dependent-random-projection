```python
import os
import time
import json
import math
import copy
import random
from pathlib import Path
from typing import List, Dict, Tuple

import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
from torchvision import datasets, models
from torchvision.utils import make_grid

try:
    from tqdm import tqdm
except Exception:
    import sys
    !{sys.executable} -m pip install tqdm
    from tqdm import tqdm

# --- Device selection (CUDA > MPS > CPU)
if torch.cuda.is_available():
    DEVICE = torch.device("cuda")
elif torch.backends.mps.is_available() and torch.backends.mps.is_built():
    DEVICE = torch.device("mps")
else:
    DEVICE = torch.device("cpu")
print("Device:", DEVICE)

# ----------------------
# Utilities, logging
# ----------------------
def ensure_dir(path):
    os.makedirs(path, exist_ok=True)

def save_json(obj, path):
    with open(path, "w") as f:
        json.dump(obj, f, indent=2)

# ----------------------
# 1) Lion optimizer (from your original)
# ----------------------
class Lion(optim.Optimizer):
    def __init__(self, params, lr=1e-4, betas=(0.9, 0.99), weight_decay=0.0):
        if not 0.0 <= lr:
            raise ValueError(f"Invalid learning rate: {lr}")
        if not 0.0 <= betas[0] < 1.0:
            raise ValueError(f"Invalid beta parameter at index 0: {betas[0]}")
        if not 0.0 <= betas[1] < 1.0:
            raise ValueError(f"Invalid beta parameter at index 1: {betas[1]}")
        defaults = dict(lr=lr, betas=betas, weight_decay=weight_decay)
        super(Lion, self).__init__(params, defaults)

    @torch.no_grad()
    def step(self, closure=None):
        loss = None
        if closure is not None:
            with torch.enable_grad():
                loss = closure()

        for group in self.param_groups:
            for p in group['params']:
                if p.grad is None:
                    continue

                if group['weight_decay'] != 0:
                    p.grad.add_(p, alpha=group['weight_decay'])

                grad = p.grad
                state = self.state[p]
                lr, (beta1, beta2) = group['lr'], group['betas']

                if len(state) == 0:
                    state['exp_avg'] = torch.zeros_like(p)

                exp_avg = state['exp_avg']

                update = exp_avg * beta1 + grad * (1 - beta1)
                p.add_(torch.sign(update), alpha=-lr)
                exp_avg.mul_(beta2).add_(grad, alpha=1 - beta2)
        return loss

# ----------------------
# 2) SAM (Sharpness-Aware Minimization) wrapper
#    (two-step optimizer wrapper)
# ----------------------
class SAM:
    def __init__(self, params, base_optimizer, rho=0.05, adaptive=False, **opt_kwargs):
        # base_optimizer is optimizer class (e.g., optim.SGD)
        self.base_optimizer = base_optimizer
        self.rho = rho
        self.adaptive = adaptive
        self.opt = base_optimizer(params, **opt_kwargs)

    @torch.no_grad()
    def first_step(self, zero_grad=True):
        grad_norm = self._grad_norm()
        for group in self.opt.param_groups:
            scale = self.rho / (grad_norm + 1e-12)
            for p in group["params"]:
                if p.grad is None: continue
                e_w = (torch.pow(p, 2) if self.adaptive else 1.0) * p.grad * scale
                p.add_(e_w)  # climb to the local maximum "on the loss landscape"
                p.state['e_w'] = e_w
        if zero_grad:
            self.opt.zero_grad()

    @torch.no_grad()
    def second_step(self, zero_grad=True):
        for group in self.opt.param_groups:
            for p in group["params"]:
                if 'e_w' in p.state:
                    p.sub_(p.state['e_w'])  # return to original weights
                    del p.state['e_w']
        self.opt.step()  # do the actual step
        if zero_grad:
            self.opt.zero_grad()

    def step(self):
        raise NotImplementedError("Use first_step() and second_step() for SAM")

    def zero_grad(self):
        self.opt.zero_grad()

    def _grad_norm(self):
        device = next(self.opt.param_groups[0]['params']).device
        shared_device = device
        norm = torch.norm(
            torch.stack([
                ((torch.abs(p) if self.adaptive else 1.0) * p.grad).norm(p=2).to(shared_device)
                for group in self.opt.param_groups for p in group['params']
                if p.grad is not None
            ]),
            p=2
        )
        return norm

# ----------------------
# 3) SimpleCNN with spectral norm switch
# ----------------------
from torch.nn.utils import spectral_norm

class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10, input_channels=3, use_spectral_norm=False):
        super(SimpleCNN, self).__init__()
        def sn(layer):
            return spectral_norm(layer) if use_spectral_norm else layer

        self.conv1 = sn(nn.Conv2d(input_channels, 32, kernel_size=3, padding=1))
        self.conv2 = sn(nn.Conv2d(32, 64, kernel_size=3, padding=1))
        self.pool1 = nn.MaxPool2d(2,2)
        self.bn1 = nn.BatchNorm2d(64)

        self.conv3 = sn(nn.Conv2d(64,128,kernel_size=3,padding=1))
        self.conv4 = sn(nn.Conv2d(128,128,kernel_size=3,padding=1))
        self.pool2 = nn.MaxPool2d(2,2)
        self.bn2 = nn.BatchNorm2d(128)

        if input_channels == 1:
            self.fc_input_features = 128 * 7 * 7
        else:
            self.fc_input_features = 128 * 8 * 8

        self.fc1 = sn(nn.Linear(self.fc_input_features, 512))
        self.dropout = nn.Dropout(0.5)
        self.fc2 = sn(nn.Linear(512, num_classes))

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = self.pool1(x)
        x = self.bn1(x)

        x = F.relu(self.conv3(x))
        x = F.relu(self.conv4(x))
        x = self.pool2(x)
        x = self.bn2(x)

        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

# ----------------------
# 4) Apply spectral normalization to an arbitrary model (ResNet)
# ----------------------
def apply_spectral_norm_model(model: nn.Module, use_sn: bool):
    """
    Recursively wraps Conv2d and Linear layers with spectral_norm (or removes it).
    If use_sn is False, returns model unchanged (no SN applied).
    """
    if not use_sn:
        return model

    for name, module in model.named_children():
        if isinstance(module, (nn.Conv2d, nn.Linear)):
            wrapped = spectral_norm(module)
            setattr(model, name, wrapped)
        else:
            apply_spectral_norm_model(module, use_sn)
    return model

# ----------------------
# 5) Spectral norm estimation per layer (power iteration)
# ----------------------
@torch.no_grad()
def estimate_spectral_norms(model: nn.Module, n_iters: int = 20, device: torch.device = DEVICE) -> Dict[str, float]:
    """
    Estimate the spectral norm (largest singular value) for each Conv2d / Linear weight.
    Returns dict: {param_name: spectral_norm}
    """
    sn_dict = {}
    model_cpu = model.to("cpu")  # operate on cpu for consistent numeric operations
    for name, module in model_cpu.named_modules():
        if isinstance(module, (nn.Conv2d, nn.Linear)):
            W = module.weight.data.detach().reshape(module.weight.shape[0], -1).float().cpu().numpy()
            # power iteration in numpy
            v = np.random.randn(W.shape[1])
            v = v / (np.linalg.norm(v) + 1e-12)
            for _ in range(n_iters):
                u = W.dot(v)
                u_norm = np.linalg.norm(u) + 1e-12
                u = u / u_norm
                v = W.T.dot(u)
                v = v / (np.linalg.norm(v) + 1e-12)
            sigma = float(u.dot(W.dot(v)))
            sn_dict[f"{name}.weight"] = abs(sigma)
    model.to(device)
    return sn_dict



# ----------------------
# 6) Gradient penalty (regularize gradient of logits w.r.t. inputs)
#    simple L2 penalty on gradient norm per sample minus target scalar
# ----------------------
def gradient_penalty(model: nn.Module, inputs: torch.Tensor, targets: torch.Tensor, device: torch.device, lp: float = 2.0):
    """
    Computes average gradient norm (L2) of the model outputs (logits) wrt inputs.
    Returns penalty scalar to add to loss (mean over batch).
    """
    inputs = inputs.clone().detach().to(device).requires_grad_(True)
    outputs = model(inputs)
    # Choose scalar to differentiate: sum of logits of the true class
    # (could also use softmax or full jacobian norm; this is a simple workable proxy)
    chosen = outputs.gather(1, targets.view(-1,1)).squeeze()
    grads = torch.autograd.grad(chosen.sum(), inputs, create_graph=True)[0]
    grad_norms = grads.view(grads.size(0), -1).norm(p=lp, dim=1)
    return grad_norms.mean()



# 7) Data loaders (MNIST, CIFAR-10, CIFAR-100, TinyImageNet)
#    NOTE: TinyImageNet is not provided by torchvision by default. Expect it to be placed at ./data/tiny-imagenet-200
# ----------------------
def get_data(dataset_name: str, batch_size: int = 128, num_workers:int=4):
    name = dataset_name.upper()
    if name == "MNIST":
        transform_train = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.1307,), (0.3081,))
        ])
        transform_test = transform_train
        trainset = datasets.MNIST(root="./data", train=True, download=True, transform=transform_train)
        testset = datasets.MNIST(root="./data", train=False, download=True, transform=transform_test)
        input_channels = 1
        num_classes = 10
    elif name == "CIFAR10":
        transform_train = transforms.Compose([
            transforms.RandomCrop(32, padding=4),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize((0.4914,0.4822,0.4465),(0.2023,0.1994,0.2010))
        ])
        transform_test = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.4914,0.4822,0.4465),(0.2023,0.1994,0.2010))
        ])
        trainset = datasets.CIFAR10(root="./data", train=True, download=True, transform=transform_train)
        testset = datasets.CIFAR10(root="./data", train=False, download=True, transform=transform_test)
        input_channels = 3
        num_classes = 10
    elif name == "CIFAR100":
        transform_train = transforms.Compose([
            transforms.RandomCrop(32, padding=4),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize((0.5071,0.4867,0.4408),(0.2675,0.2565,0.2761))
        ])
        transform_test = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.5071,0.4867,0.4408),(0.2675,0.2565,0.2761))
        ])
        trainset = datasets.CIFAR100(root="./data", train=True, download=True, transform=transform_train)
        testset = datasets.CIFAR100(root="./data", train=False, download=True, transform=transform_test)
        input_channels = 3
        num_classes = 100
    elif name == "TINYIMAGENET":
        # Expect dataset placed at ./data/tiny-imagenet-200 (standard layout)
        tiny_dir = Path("./data/tiny-imagenet-200")
        if not tiny_dir.exists():
            raise FileNotFoundError(
                "TinyImageNet not found at ./data/tiny-imagenet-200.\n"
                "Please download TinyImageNet (200 classes) and place it under ./data/tiny-imagenet-200\n"
                "Typical structure: ./data/tiny-imagenet-200/train/<class>/images/..., ./data/tiny-imagenet-200/val/..., ./data/tiny-imagenet-200/test/...\n"
                "Many tutorial download links exist (Stanford TinyImageNet or Kaggle)."
            )
        transform_train = transforms.Compose([
            transforms.RandomResizedCrop(64),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize((0.4802,0.4481,0.3975),(0.2302,0.2265,0.2262))
        ])
        transform_test = transforms.Compose([
            transforms.Resize(64),
            transforms.CenterCrop(64),
            transforms.ToTensor(),
            transforms.Normalize((0.4802,0.4481,0.3975),(0.2302,0.2265,0.2262))
        ])
        trainset = torchvision.datasets.ImageFolder(root=str(tiny_dir / "train"), transform=transform_train)
        testset = torchvision.datasets.ImageFolder(root=str(tiny_dir / "val"), transform=transform_test)
        input_channels = 3
        num_classes = 200
    else:
        raise ValueError(f"Unknown dataset {dataset_name}")

    trainloader = DataLoader(trainset, batch_size=batch_size, shuffle=True, num_workers=num_workers, pin_memory=(DEVICE.type == "cuda"))
    testloader = DataLoader(testset, batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=(DEVICE.type == "cuda"))
    return trainloader, testloader, input_channels, num_classes

# ----------------------
# 8) Metrics helpers
# ----------------------
def get_grad_norm(model: nn.Module):
    total_norm = 0.0
    for p in model.parameters():
        if p.grad is not None:
            param_norm = p.grad.data.norm(2)
            total_norm += param_norm.item() ** 2
    return total_norm ** 0.5

def get_param_displacement(params0: List[torch.Tensor], params1: List[torch.Tensor]):
    displacement = 0.0
    for p0, p1 in zip(params0, params1):
        displacement += torch.norm(p0 - p1).item() ** 2
    return math.sqrt(displacement)

# ----------------------
# 9) Training and evaluation loops supporting:
#    - spectral normalization flag (applied to model before)
#    - gradient penalty weight
#    - choice of optimizer: sgd | adam | adamw | lion | sam_sgd | sam_adam
# ----------------------
def build_optimizer(name: str, model_params, lr: float, weight_decay: float):
    name = name.lower()
    if name == 'sgd':
        return optim.SGD(model_params, lr=lr, momentum=0.9, weight_decay=weight_decay)
    elif name == 'adam':
        return optim.Adam(model_params, lr=lr, weight_decay=weight_decay)
    elif name == 'adamw':
        return optim.AdamW(model_params, lr=lr, weight_decay=weight_decay)
    elif name == 'lion':
        return Lion(model_params, lr=lr, weight_decay=weight_decay)
    else:
        raise ValueError(f"Unsupported optimizer {name}")

def train_epoch(model: nn.Module, trainloader: DataLoader, optimizer, criterion, device: torch.device,
                gradient_penalty_weight: float = 0.0, use_sam: bool = False, sam_wrapper: SAM = None):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    grad_norms = []
    for inputs, targets in trainloader:
        inputs, targets = inputs.to(device), targets.to(device)

        if use_sam:
            # SAM two-step procedure
            # 1. compute gradients
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            if gradient_penalty_weight > 0.0:
                gp = gradient_penalty(model, inputs, targets, device)
                loss = loss + gradient_penalty_weight * gp
            loss.backward()
            sam_wrapper.first_step(zero_grad=True)

            # 2. second forward-backward step
            outputs2 = model(inputs)
            loss2 = criterion(outputs2, targets)
            if gradient_penalty_weight > 0.0:
                gp2 = gradient_penalty(model, inputs, targets, device)
                loss2 = loss2 + gradient_penalty_weight * gp2
            loss2.backward()
            sam_wrapper.second_step(zero_grad=True)

            grad_norms.append(get_grad_norm(model))
            running_loss += float(loss2.item()) * inputs.size(0)
            _, predicted = outputs2.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
        else:
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            if gradient_penalty_weight > 0.0:
                gp = gradient_penalty(model, inputs, targets, device)
                loss = loss + gradient_penalty_weight * gp
            loss.backward()
            grad_norms.append(get_grad_norm(model))
            optimizer.step()

            running_loss += float(loss.item()) * inputs.size(0)
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()

    avg_loss = running_loss / total
    acc = 100.0 * correct / total
    avg_grad_norm = float(np.mean(grad_norms)) if grad_norms else 0.0
    return avg_loss, acc, avg_grad_norm

def eval_epoch(model: nn.Module, testloader: DataLoader, criterion, device: torch.device):
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, targets in testloader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            running_loss += float(loss.item()) * inputs.size(0)
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
    avg_loss = running_loss / total
    acc = 100.0 * correct / total
    return avg_loss, acc

# ----------------------
# 10) Experiment runner implementing Week plan requirements
# ----------------------
def run_experiment(
    dataset: str,
    backbone: str = "simplecnn",     # simplecnn or resnet18
    use_spectral_norm: bool = False,
    gradient_penalty_weight: float = 0.0,
    optimizer_name: str = "sgd",
    use_sam: bool = False,
    sam_params: dict = None,
    lr: float = 1e-3,
    weight_decay: float = 1e-4,
    batch_size: int = 128,
    epochs: int = 10,
    results_dir: str = "results"
):
    ensure_dir(results_dir)
    run_tag = f"{dataset}_{backbone}_{optimizer_name}_SN{int(use_spectral_norm)}_GP{gradient_penalty_weight}_SAM{int(use_sam)}"
    print(f"\n=== RUN: {run_tag} ===")

    trainloader, testloader, input_channels, num_classes = get_data(dataset, batch_size=batch_size)

    # Build model
    if backbone.lower() == "simplecnn":
        model = SimpleCNN(num_classes=num_classes, input_channels=input_channels, use_spectral_norm=False)
    elif backbone.lower() == "resnet18":
        # Use new torchvision weights argument to avoid deprecation warning
        from torchvision.models import ResNet18_Weights
        model = models.resnet18(weights=None)  # equivalent to pretrained=False
        # adapt final layer
        model.fc = nn.Linear(model.fc.in_features, num_classes)
    else:
        raise ValueError("Unknown backbone")

    # Optionally apply spectral norm
    if use_spectral_norm:
        model = apply_spectral_norm_model(model, use_sn=True)

    model = model.to(DEVICE)
    criterion = nn.CrossEntropyLoss()

    # optimizer / SAM
    if use_sam:
        base_opt = optim.SGD if optimizer_name.lower() in ['sgd'] else optim.Adam
        sam_params = sam_params or {'rho': 0.05, 'adaptive': False}
        if base_opt is optim.SGD:
            base_opt_args = {'lr': lr, 'momentum': 0.9, 'weight_decay': weight_decay}
        else:
            base_opt_args = {'lr': lr, 'weight_decay': weight_decay}
        sam_wrapper = SAM(model.parameters(), base_optimizer=base_opt, rho=sam_params.get('rho',0.05),
                          adaptive=sam_params.get('adaptive', False), **base_opt_args)
        optimizer = sam_wrapper
    else:
        optimizer = build_optimizer(optimizer_name, model.parameters(), lr=lr, weight_decay=weight_decay)
        sam_wrapper = None

    # initial parameters snapshot for displacement
    initial_params = [p.clone().detach().cpu() for p in model.parameters()]

    logs = {
        'train_loss': [], 'train_acc': [], 'train_grad_norm': [],
        'test_loss': [], 'test_acc': [],
        'param_displacement': [], 'spectral_norms': []
    }

    for epoch in range(epochs):
        start = time.time()

        if use_sam:
            train_loss, train_acc, avg_grad_norm = train_epoch(
                model, trainloader, optimizer=None, criterion=criterion, device=DEVICE,
                gradient_penalty_weight=gradient_penalty_weight, use_sam=True, sam_wrapper=sam_wrapper
            )
        else:
            train_loss, train_acc, avg_grad_norm = train_epoch(
                model, trainloader, optimizer=optimizer, criterion=criterion, device=DEVICE,
                gradient_penalty_weight=gradient_penalty_weight, use_sam=False, sam_wrapper=None
            )

        test_loss, test_acc = eval_epoch(model, testloader, criterion, device=DEVICE)

        # spectral norms per layer
        sn_per_layer = estimate_spectral_norms(model, n_iters=10, device=DEVICE)

        # param displacement
        current_params = [p.clone().detach().cpu() for p in model.parameters()]
        displacement = get_param_displacement(initial_params, current_params)

        # log metrics
        logs['train_loss'].append(train_loss)
        logs['train_acc'].append(train_acc)
        logs['train_grad_norm'].append(avg_grad_norm)
        logs['test_loss'].append(test_loss)
        logs['test_acc'].append(test_acc)
        logs['param_displacement'].append(displacement)
        logs['spectral_norms'].append(sn_per_layer)

        dur = time.time() - start
        print(f"Epoch {epoch+1}/{epochs} | train_loss={train_loss:.4f} train_acc={train_acc:.2f}% | "
              f"test_loss={test_loss:.4f} test_acc={test_acc:.2f}% | grad_norm={avg_grad_norm:.4f} | "
              f"displacement={displacement:.4f} | time={dur:.1f}s")

        # save per-epoch logs
        save_json(logs, os.path.join(results_dir, f"{run_tag}_logs.json"))

    print("Experiment finished. Saved logs to results directory.")
    return logs, model

# ----------------------
# 11) Simple plotting utilities for per-layer spectral norm evolution
# ----------------------
def plot_accuracy_loss(logs, title="Experiment"):
    epochs = range(1, len(logs['train_loss']) + 1)
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(epochs, logs['train_loss'], '-o', label='train_loss')
    plt.plot(epochs, logs['test_loss'], '-o', label='test_loss')
    plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.title(f"{title} - Loss"); plt.legend()
    plt.grid(True)

    plt.subplot(1,2,2)
    plt.plot(epochs, logs['train_acc'], '-o', label='train_acc')
    plt.plot(epochs, logs['test_acc'], '-o', label='test_acc')
    plt.xlabel('Epoch'); plt.ylabel('Accuracy (%)'); plt.title(f"{title} - Accuracy"); plt.legend()
    plt.grid(True)
    plt.show()

def plot_spectral_norms(logs, layer_name=None, top_k=6):
    # logs['spectral_norms'] is a list of dicts per epoch {param_name: sigma}
    epochs = len(logs['spectral_norms'])
    if epochs == 0:
        print("No spectral norm data.")
        return
    # pick layers present in first epoch
    first = logs['spectral_norms'][0]
    layer_names = sorted(first.keys())
    if layer_name:
        chosen = [layer_name] if layer_name in layer_names else []
    else:
        # choose top_k param names by avg sigma
        avg_sigma = {name: np.mean([epoch.get(name, 0.0) for epoch in logs['spectral_norms']]) for name in layer_names}
        chosen = sorted(avg_sigma.keys(), key=lambda k: avg_sigma[k], reverse=True)[:top_k]

    plt.figure(figsize=(10,5))
    for name in chosen:
        vals = [epoch.get(name, 0.0) for epoch in logs['spectral_norms']]
        plt.plot(range(1, epochs+1), vals, '-o', label=name)
    plt.xlabel('Epoch'); plt.ylabel('Spectral Norm'); plt.title('Spectral Norm Evolution'); plt.legend(loc='upper right'); plt.grid(True)
    plt.show()

# 12) Lipschitz constant plotting
# ----------------------
def plot_lipschitz(logs, title="Lipschitz Constant"):
    if 'lipschitz' not in logs or len(logs['lipschitz']) == 0:
        print("No Lipschitz data.")
        return
    plt.figure(figsize=(8,4))
    plt.plot(range(1, len(logs['lipschitz'])+1), logs['lipschitz'], '-o', label='Lipschitz (max SN)')
    plt.xlabel('Epoch')
    plt.ylabel('Lipschitz estimate')
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.show()


# ----------------------
# Experiments
# ----------------------

# -----------------------------
# Compare optimizers with/without spectral normalization
# -----------------------------

optimizers = [
    {"name": "sgd", "lr": 0.01},
    {"name": "adamw", "lr": 0.001},
    {"name": "lion", "lr": 1e-4},
]

use_sam_options = {
    "sgd": False,
    "adamw": False,
    "lion": False,
}

gradient_penalty_options = {
    "sgd": 0.0,
    "adamw": 1e-3,  # you can adjust per optimizer if desired
    "lion": 0.0,
}

epochs = 20
batch_size = 128

results = {}

for opt in optimizers:
    optimizer_name = opt["name"]
    lr = opt["lr"]
    print(f"\n=== Running optimizer: {optimizer_name} ===")

    results[optimizer_name] = {}

    for use_sn in [False, True]:
        sn_flag = "with_SN" if use_sn else "no_SN"
        print(f"\n--- Spectral Norm: {sn_flag} ---")

        logs, model = run_experiment(
            dataset="CIFAR10",
            backbone="SimpleCNN",
            use_spectral_norm=use_sn,
            gradient_penalty_weight=gradient_penalty_options.get(optimizer_name, 0.0),
            optimizer_name=optimizer_name,
            use_sam=use_sam_options.get(optimizer_name, False),
            sam_params={'rho':0.05,'adaptive':False} if use_sam_options.get(optimizer_name, False) else None,
            lr=lr,
            weight_decay=1e-4,
            batch_size=batch_size,
            epochs=epochs
        )

        results[optimizer_name][sn_flag] = {"logs": logs, "model": model}

        # Plot accuracy/loss
        plot_accuracy_loss(logs, title=f"CIFAR10 SimpleCNN{optimizer_name.upper()} {sn_flag}")
        # Plot spectral norms
        plot_spectral_norms(logs, top_k=6)
        # Plot Lipschitz if you added it
        if 'lipschitz' in logs:
            plot_lipschitz(logs, title=f"Lipschitz {optimizer_name.upper()} {sn_flag}")
```
